### There are 2 main rules that performing matrix multiplication need need to satisfy
1. The **inner dimensions** must match
* `(3,2) @ (3,2)` won't work
* `(2,3) @ (3,2)` will work
* `(3,2) @ (2,3)` will work
2. The resulting matrix has the shape of the  **outer dimensions**
 `(2,3) @ (3,2)` -> `(2,2)`

```python
torch.matmul(torch.rand(10,2), torch.rand(2,10)).shape
```

### Reshaing, stacking, squeezing and unsqueezing tensors

* Reshaping - reshape an input input tensor to a defined shape
* View -Return a view of an inpput tensor of certain shape but keep the same memory
* Stacking -Combine multiple tensors on the top of each other (vstack) or side by side (hstack)
* Squeeze - removes all `1` dimensions from a tensor
* Unsqueez - add a `1` dimension to a target tensor
* Permute -Return a view of the input with dimensions permuted (swapped ) in a certain way

```python
#Lets create a tensor
x = torch.arange(1., 9.)
x, x.shape
```

```python
 # Add an extra dimension
x_reshaped = x.reshape(4,2)
x_reshaped, x_reshaped.shape
```

```python
# Change the view
z = x.view(4,2)
z, z.shape
```

```python
# Changing z changes x (because a view of a ensor shares the same memory as the original input)
z[:, 0] = 5
z, x
```

```python
# Stack tensors on top each other
x_stacked = torch.stack([x, x, x, x], dim = 1)
x_stacked, x_stacked.shape
```

## Reproductibility (Trying to take random out of random)
 In short how a neural network learns:
  `start with random numbers -> tensor operations -> update random numbers to try and make them better representations of the data -> again -> again -> and again ...`

  To reduce the randomness in neural networks and PyTorch comes the concept of a  **random seed**
  Essentially what the random seed does is "flavour" the randomness.

```python
torch.rand(3,3)
```

```python
 # Create two random tensors
random_tensor_A = torch.rand(3,4)
random_tensor_B = torch.rand(3,4)

print(random_tensor_A)
print(random_tensor_B)
print(random_tensor_A == random_tensor_B)
```

```python
# Lets make some random but reproducible tenosrs
RANDOM_SEED = 42

torch.manual_seed(RANDOM_SEED)
random_tensor_C = torch.rand(3,4)
torch.manual_seed(RANDOM_SEED)
random_tensor_D = torch.rand(3,4)

print(random_tensor_C)
print(random_tensor_D)
print(random_tensor_C == random_tensor_D)
```
